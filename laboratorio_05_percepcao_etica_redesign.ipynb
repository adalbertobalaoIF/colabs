# Laboratório 05 — Percepção Visual, Ética e Redesign
Objetivo: identificar problemas visuais e reconstruir um gráfico com clareza e integridade.


```python
import pandas as pd
import matplotlib.pyplot as plt

categorias = ['A','B','C','D']
valores = [92, 96, 94, 99]

plt.figure(figsize=(8,4))
plt.bar(categorias, valores)
plt.ylim(90,100)
plt.title('Gráfico propositalmente problemático')
plt.show()
```

## Análise crítica
Identifique problemas de escala, destaque, contexto, título, cores e mensagem.

```python
plt.figure(figsize=(8,4))
plt.bar(categorias, valores)
plt.ylim(0,105)
plt.title('Valores por categoria')
plt.ylabel('Valor')
plt.grid(axis='y', alpha=.2)
plt.show()
```
