Objetivo: transformar uma demanda genérica em perguntas analíticas e produtos visuais orientados a públicos distintos.


## Situação-problema
Uma instituição deseja compreender matrículas e desempenho acadêmico. Defina três públicos e formule perguntas adequadas a cada um.

````python
import pandas as pd

briefing = pd.DataFrame({
    'publico': ['Gestão', 'Coordenação', 'Estudantes'],
    'decisao': [
        'Alocação de recursos, planejamento estratégico, avaliação da saúde institucional.',
        'Ajustes de currículo, suporte a grupos específicos de estudantes, planejamento de ofertas de cursos.',
        'Seleção de cursos, estratégias de estudo, identificação de recursos de apoio acadêmico.'
    ],
    'pergunta_analitica': [
        'Qual a tendência geral do desempenho acadêmico e da retenção de estudantes na instituição nos últimos 5 anos?',
        'Quais disciplinas ou departamentos apresentam maior/menor taxa de sucesso estudantil e evasão?',
        'Como meu desempenho acadêmico se compara ao dos meus colegas em disciplinas e programas semelhantes?'
    ],
    'indicador_principal': [
        'Taxa de Retenção Anual, Taxa de Conclusão de Curso, Média Geral de Notas (GPA) Institucional.',
        'Taxa de Sucesso por Disciplina/Departamento, Taxa de Evasão por Disciplina/Departamento, Avaliação de Disciplina pelos Estudantes.',
        'Média Individual de Notas (GPA) vs. Média da Turma/Curso, Taxa de Aprovação em Disciplinas Pré-Requisito, Posição no Ranking da Turma.'
    ],
    'mensagem_esperada': [
        'Visão estratégica da eficácia institucional e áreas para investimento ou intervenção.',
        'Identificação de pontos fortes e fracos em cursos e departamentos para otimizar o suporte e o ensino.',
        'Autoavaliação do progresso acadêmico e direcionamento para melhoria e uso de recursos.'
    ]
})
briefing
````
