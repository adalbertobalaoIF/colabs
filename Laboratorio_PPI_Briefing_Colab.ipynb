# PARTE E — VALIDAÇÃO DO BRIEFING

Execute a célula abaixo para verificar se o grupo deixou campos essenciais vazios ou se há inconsistências básicas.

```python
# Validação automática do briefing

campos_essenciais = {
    "Problema público": problema_publico,
    "Público-alvo": publico_alvo,
    "Decisão apoiada": decisao_apoiada,
    "Pergunta analítica": pergunta_analitica,
    "Variáveis de entrada": variaveis_entrada,
    "Fontes de dados": fontes_dados,
    "Granularidade": granularidade or exemplo_registro,
    "Telas do dashboard": telas,
    "KPIs": kpis,
}

problemas = []

for campo, valor in campos_essenciais.items():
    if not str(valor).strip():
        problemas.append(f"❌ Campo não preenchido: {campo}")

if tipo_ml != "Não se aplica":
    if not o_que_sera_previsto.strip():
        problemas.append("❌ Foi selecionado ML, mas não foi definido o que será previsto/classificado/agrupado.")
    if not metrica_avaliacao.strip():
        problemas.append("⚠️ Foi selecionado ML, mas ainda não foi indicada uma métrica de avaliação.")

if usa_otimizacao == "Sim":
    if not recurso_limitado.strip():
        problemas.append("❌ Há otimização, mas o recurso limitado não foi definido.")
    if not objetivo_otimizacao.strip():
        problemas.append("❌ Há otimização, mas o objetivo não foi definido.")

if len(problema_publico.strip()) < 25:
    problemas.append("⚠️ O problema público parece muito curto. Tornem a dor mais específica.")

if len(pergunta_analitica.strip()) < 15:
    problemas.append("⚠️ A pergunta analítica parece genérica ou curta.")

print("=" * 70)
print("DIAGNÓSTICO DO BRIEFING")
print("=" * 70)

if problemas:
    for p in problemas:
        print(p)
    print(f"\nTotal de observações: {len(problemas)}")
else:
    print("✅ Briefing preenchido sem pendências básicas.")
```
