# Laboratório 04 — Agregação e Valores Multiagregados
Objetivo: construir indicadores com GroupBy, agg, pivot_table e proporções.


```python
import pandas as pd
import seaborn as sns

dados = sns.load_dataset('tips')
dados.head()
```

```python
indicadores = dados.groupby('day').agg(
    atendimentos=('total_bill','count'),
    faturamento=('total_bill','sum'),
    ticket_medio=('total_bill','mean'),
    gorjeta_media=('tip','mean')
).round(2)
indicadores
```

```python
tabela = pd.pivot_table(dados, index='day', columns='time', values='total_bill', aggfunc=['count','mean'])
tabela
```

## Desafio
Compare valores absolutos e relativos e explique qual nível de agregação responde melhor à pergunta.

Para responder à pergunta de negócio de forma precisa, a análise entre valores absolutos e relativos precisa ser separada pelo nível de agregação utilizado:

1. Agregação Geral por Dia (groupby('day'))

Absoluto (faturamento / atendimentos): Mostra o volume bruto. Sábado gera a maior receita total (1.778,40) e número de atendimentos (87), enquanto a Sexta-feira tem o menor faturamento (325,88).

Relativo (ticket_medio / gorjeta_media): Revela a eficiência por cliente. O Domingo apresenta o maior ticket médio (21,41) e a maior gorjeta média (3,26), superando o Sábado em rentabilidade por mesa.

2. Agregação Multivariada por Dia e Horário (pivot_table)

Absoluto (count): Evidencia que a operação de Quinta-feira é quase exclusivamente no Lunch (61 de 62 atendimentos), enquanto Sábado e Domingo operam 100% no Dinner.

Relativo (mean): Demonstra que os tickets médios do Dinner em todos os dias (variando de 18,78 a 21,41) são consistentemente superiores aos do Lunch (Sexta: 12,85; Quinta: 17,66).

Qual nível de agregação responde melhor à pergunta?

A escolha do nível ideal depende da pergunta específica de negócio:

Para decisões de escala e estoque (Agregação Simples por Dia): Responde melhor sobre a demanda geral e a alocação global de recursos ao longo da semana.

Para decisões operacionais, pricing e escala de equipe (Agregação Multivariada via pivot_table): É o nível superior e mais adequado. O agrupamento por dia e horário revela a forte polarização entre o público de almoço e jantar — um padrão invisível quando analisamos apenas o dia consolidado. Por exemplo, a Sexta-feira parece fraca no acumulado diário, mas seu jantar atinge um ticket médio de 19,66, próximo aos finais de semana.
