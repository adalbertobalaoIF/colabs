# Laboratório 04 — Distâncias e Similaridades

**Base:** Wine  
**Tempo estimado:** 60–75 minutos

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from scipy.spatial.distance import euclidean, cityblock, cosine

wine = load_wine(as_frame=True)
df = wine.frame.copy()
df.head()
```

## 1. Seleção de observações

```python
atributos = ["alcohol", "malic_acid", "color_intensity", "proline"]
amostra = df.loc[[0, 1, 50], atributos]
amostra
```

## 2. Distâncias sem escalonamento

```python
a = amostra.iloc[0].to_numpy()
b = amostra.iloc[1].to_numpy()

resultados_brutos = {
    "Euclidiana": euclidean(a, b),
    "Manhattan": cityblock(a, b),
    "Cosseno (distância)": cosine(a, b)
}
resultados_brutos
```

## 3. Padronização

```python
scaler = StandardScaler()
df_pad = pd.DataFrame(
    scaler.fit_transform(df[atributos]),
    columns=atributos,
    index=df.index
)

a_pad = df_pad.loc[0].to_numpy()
b_pad = df_pad.loc[1].to_numpy()

resultados_padronizados = {
    "Euclidiana": euclidean(a_pad, b_pad),
    "Manhattan": cityblock(a_pad, b_pad),
    "Cosseno (distância)": cosine(a_pad, b_pad)
}
resultados_padronizados
```

## 4. Comparação

```python
comparacao = pd.DataFrame({
    "Sem escalonamento": resultados_brutos,
    "Após padronização": resultados_padronizados
})
comparacao
```
