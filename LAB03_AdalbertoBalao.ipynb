# Laboratório 03 — Limpeza e Transformação
Objetivo: tratar valores ausentes, padronizar categorias e criar novas variáveis.


```python
import pandas as pd
import seaborn as sns

dados = sns.load_dataset('titanic')
dados.head()
```

```python
dados.isna().sum().sort_values(ascending=False).head(10)
```

```python
dados_limpos = dados.copy()
dados_limpos['age'] = dados_limpos['age'].fillna(dados_limpos['age'].median())
dados_limpos['embarked'] = dados_limpos['embarked'].fillna(dados_limpos['embarked'].mode()[0])
dados_limpos = dados_limpos.drop_duplicates()
dados_limpos['faixa_etaria'] = pd.cut(dados_limpos['age'], bins=[0,12,18,35,60,100], labels=['Criança','Adolescente','Adulto jovem','Adulto','Idoso'])
dados_limpos[['age','faixa_etaria','embarked']].head()
```

## Desafio
Documente cada decisão de limpeza e explique como ela pode afetar as visualizações.

1. Imputação de Nulos em age pela Mediana

Decisão: Substituição de valores ausentes na coluna age pelo valor mediano dos dados existentes.

Impacto nas Visualizações:

Histogramas e Boxplots: Cria um pico desproporcional na barra correspondente à mediana, reduzindo artificialmente a variabilidade (desvio padrão) e achatando os quartis.

Gráficos de Dispersão (Scatter Plots): Gera um alinhamento artificial de pontos ao longo da linha da mediana no eixo correspondente à idade, o que pode mascarar correlações reais com outras variáveis.

2. Imputação de Nulos em embarked pela Moda

Decisão: Preenchimento de valores ausentes na variável categórica embarked com a categoria mais frequente (mode()[0]).

Impacto nas Visualizações:

Gráficos de Barras / Contagem: Aumenta ligeiramente a contagem do porto mais comum (Southampton - 'S'), sem alterar significativamente a proporção visual geral, evitando que a categoria seja omitida ou tratada como "Desconhecido".

3. Remoção de Linhas Duplicadas (drop_duplicates)

Decisão: Exclusão de registros completamente idênticos na base de dados.

Impacto nas Visualizações:

Gráficos de Métrica Geral (Totais/Contagens): Previne a insuflação artificial das contagens totais e evita distorções de proporção em gráficos de pizza, rosca ou barras empilhadas causadas pela dupla contagem de um mesmo indivíduo.

4. Categorização/Discretização de age em faixa_etaria

Decisão: Agrupamento da variável contínua age em intervalos definidos ([0,12,18,35,60,100]) com rótulos amigáveis.

Impacto nas Visualizações:

Facilidade de Agrupamento: Transforma uma métrica contínua em uma dimensão clara, permitindo criar gráficos de barras comparativos (ex.: taxa de sobrevivência por Criança, Adultez, etc.) e cruzamentos em mapas de calor (heatmaps).

Perda de Granularidade: Oculta detalhes ou padrões finos entre idades específicas dentro de um mesmo intervalo.
