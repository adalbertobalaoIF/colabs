```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("penguins")
df.head()
```

## 1. Diagnóstico inicial

```python
print(df.shape)
display(df.info())
display(df.isna().sum())
```

```python
dados = df.dropna().copy()
dados.describe(include="all").T
```

## 2. Distribuição univariada

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=dados, x="flipper_length_mm", hue="species", kde=True)
plt.title("Distribuição do comprimento da nadadeira")
plt.show()
```

**Interpretação:**

-

## 3. Identificação de valores extremos

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=dados, x="species", y="body_mass_g")
plt.title("Massa corporal por espécie")
plt.show()
```

**Interpretação:**

-

## 4. Relação entre variáveis

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=dados,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    style="sex"
)
plt.title("Comprimento e profundidade do bico")
plt.show()
```

**Interpretação:**

-

## 5. Correlação

```python
numericas = dados.select_dtypes(include="number")
correlacao = numericas.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(correlacao, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Matriz de correlação")
plt.show()
```
